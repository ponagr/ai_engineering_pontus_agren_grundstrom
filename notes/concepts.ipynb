## supervised learning (regression & classification)
best used for tabular data

#### regression:
- models: linear regression + knn
- metrics för evaluation: MSE, RMSE, MAE

#### classification:
- models: logistic regression + knn
- metrics för evaluation: confusion matrix, accuracy, recall, precision, f1-score

#### data splitting:
- train|test split - blandar och delar upp dataset
- train|val|test split - samma som ovan, men har en valideringsdel som används för att tuna hyperparametrar(tex k i knn med elbow plot)
    - train|test på all data, sedan en till train|test på train datan som blir x_train_val, x_val, y_train_val, y_val
        1. elbow plot på train_val
        2. välj värde på k
        3. träna på original train data
        4. evaluera på test data
        5. träna på all data
        6. export model for deployment
