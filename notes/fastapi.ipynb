# FastAPI

a python library to create APIs really simple and fast

web API is a layer or an interface to facilitate communication between frontend and backend

### CRUD(Create, Read, Update, Delete) is a simple pattern for building data APIs
- POST | Create | adds new data
- GET | Read | get the data
- PUT | Update | change an existing data
- DELETE | Delete | remove data

## create data with 'post' request (Create)

```py
# make this endpoint a post request
@app.post("/books/create_book")
async def create_book(book_request: Book): # Book = pydantic model
    new_book = Book.model_validate(book_request)
    books.append(new_book)

    return new_book # return the object that is posted
```

## update data with 'put' request (Update)

```py
# make this endpoint a put request
@app.put("/books/update_book")
async def update_book(updated_book: Book):
    for i, book in enumerate(books):
        if book.id == updated_book.id:
            books[i] = updated_book # check if id for updated book matches, then update that book
    return updated_book
```

## remove data with 'delete' request (Delete)

```py
# {id} = path parameter - value used in the function
@app.delete("/books/delete_book/{id}")
async def delete_book(id: int):
    for i, book in enumerate(books):
        if book.id == id:   # delete book if id in path parameter matches book id
            del books[i]    # example endpoint: /books/delete_book/5 -> deletes book with id 5
            break
```

## query parameter for key-value pairs in the url

```py
@app.get("/books/")
async def filter_books(
        start_year: int = Query(    # query parameter with validation and description
            1950,
            gt=1500,
            lt=CURRENT_YEAR +1,
            description="Filter books that are newer than this year"), 
        author: str = Query(
            None, 
            description="Authors firstname and lastname "),):

    # filters data to be returned based on query parameters
    filtered_books = [book for book in books if start_year < book.year]

    if author:
        filtered_books = [book for book in filtered_books if author.casefold() == book.author.casefold()]
    
    return filtered_books
```
