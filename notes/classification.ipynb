# Logistic Regression (to classify data), att predicta grupper
When we have labels we are dealing with supervised learning. 
- Lineär regression kan ha categorical features men INTE categorical LABELS (Y)

Categorical labels -> Då behöver vi en classification algoritm

Categorical features behöver encoding.      
one-hot encoding - kodar om objekt till siffror, tex kanin, hare, fisk blir 100, 010, 001.      
dummy encoding(1 dimension mindre, men lika mycket information) - kanin blir 10, hare blir 01, och blir siffran 00 är det fisk då det inte är kanin eller hare.     

modellerar probabilities för en categorical label med hjälp av logistic regression      
logistic function - s-curve -KOLLA FORMELN      
logistic regression går mellan 0 och 1, är värdet mellan 0-0.5 så räknas det som 0, och är det mellan 0.5-1 så räknas det som 1     

evaluate a classification model     
confusion matrix    

predicted vs actual     
1 vs 1 = True Positive(TP)      
0 vs 1 = False Negative(FN)     
1 vs 0 = False Positive(FP)     
0 vs 0 = True Negative(TN)      

man är ute efter högt antal TP och TN

räkna ut från confusion matrix

accuracy = (antalet TP + antalet TN) / totalt antal(TN+TP+FP+FN)
man bör inte lite på denna model endast för att den har 97% accuracy - dåligt för obalanserade datasets

precision = antal TP / antal TP + antal FP
bra när vi har låg tolerans för FP 

recall = antal TP / antal TP + antal FN
bra för låg tolerans för FN

# Random Forest/Desision Trees

no need to scale data

## desision trees
gives better results for deep learning in many situations
you divide data into different regions (R1, R2, R3)
average of the region 

### regression tree
building a tree
divide features into regions
the goal is to find regions that minimize RSS(residual sum of squares)
- recursive binary splitting:
    - find the predictor and cutoff value to minimize RSS (greedy)

minimize complexity to reduce overfitting
increase bias and decrease variance (bias/variance trade-off)
grow large tree
"prune" the tree to get subtrees
pick tree that gives you best result through crossvalidation

### classification tree
similar to regression tree but use gini index(G instead of RSS)
G measures the nodes purity(low G = high purity)
you want to minimize G

## random forest/XGBoost
improves the performance of decision trees by combining many trees using random forest and XGBoost

a decision tree has a bad predictive performance, but by combining many decision trees we can improve predictive performance

**bagging**
- sample from same dataset multiple times(bootstrapping)
- predict on each bootstrapped dataset and take average of it
- majority vote

build decision trees on bootstrapped training sets
every split in tree is based on random choice of predictors(features)

**boosting**
boost performance by growing trees based on info from previously grown trees
trees grow sequentaly(1 at a time)
each new tree attacks the residuals (errors), improving where it previously performed badly

**hyperparameters**
1. number of trees (B)- will overfit data if too many trees
2. shrinkage parameter (lambda) - controls the learning rate, learns slower if you have higher lambda
3. number of splits (d) in each tree - d=1 usualy works well

**XGBoost(Extreme Gradient Boosting)**
regularization adds into the boosting to punish weights and reduce overfitting
increase efficiency and performance

XGBoost and random forest are among the top state of the art algorithms for tabular data, can also handle missing data
