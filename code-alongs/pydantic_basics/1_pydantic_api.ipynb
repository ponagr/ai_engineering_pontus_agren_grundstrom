Pydantic is used for data validation so you can work with data in an OOP manner. Pydantic has

- type hints for schema validation
- can serialize pydantic model into json
- can deserialize json string into pydantic model

## pydantic advantages
- type safety and autocompletion - data has defined types
- input validation, so if the API changes or returns invalid data Pydantic will raise an error.
- models become self documenting with expected structure of data
- easier to test a class than json object
- can work with data in an OOP manner

## Deserialization - turn json object into Pydantic model

- data has a key: **results**
- inside of results it has value of a list of dictionaries
- another key called **count** 

we define a pydantic object by inheriting the BaseModel from pydantic and create fields which are attributes defined with type hinting. The fields have built in type validation to make sure the data follows specific schema.

- we take a json object from an API and then deserialize it to become a Python object

```python
from pydantic import BaseModel

# då vi endast skriver med results och count, så tar vi bara med dom, och skippar övriga fält som finns med i vårat api response
class PokemonListResponse(BaseModel):
    results: list[dict]
    count: int

# eftersom klassen ärver av pydantics BaseModel klass ärver vi metoden model_validate
pokemons = PokemonListResponse.model_validate(data)
pokemons
```

```python
pokemons.count
```

```python
pokemons.results
```

```python
from pydantic import computed_field

class PokemonListResponse(BaseModel):
    results: list[dict]
    count: int

    @computed_field # gör så att property får mer funktionalitet
    @property # gör så att vårat field blir en property
    def number_stored_pokemons(self) -> int:
        return len(self.results)

pokemons = PokemonListResponse.model_validate(data)

# vårat PokemonListResponse objekt har nu 3 properties, [results, count, number_stored_pokemons]
pokemons.number_stored_pokemons
```

## if we get wrong type or wrong data comes in or API changes

```python
data = requests.get("https://pokeapi.co/api/v2/pokemon-species/pikachu").json()

data.keys()
```

```python
from pydantic import ValidationError

class PokemonListResponse(BaseModel):
    results: list[dict]
    count: str

try:
    pokemons = PokemonListResponse.model_validate(data)
except ValidationError as e:
    print(e)
```

choose the types of attributes you want instead of picking all

```python
for key, values in data.items():
    print(f"{key:<30} {type(values)}")
```

```python
class Pokemon(BaseModel):
    id: int
    name: str
    base_happiness: int
    capture_rate: int
    evolution_chain: dict
    gender_rate: int
    generation: dict


pikachu = Pokemon.model_validate(data)
pikachu
```

```python
pikachu.generation
```

```python
try:
    pikachu.abilities
except AttributeError as err: 
    print(err)
```

## serialize into json
After working with an object and perhaps transforming it, you can serialize it back to a json object.   
This is particularly important in fastAPI where data is sent as serialized json objects, which is a json formatted string.  
- pydantic model -> json object

```python
serialized_pikachu = pikachu.model_dump_json()
serialized_pikachu
```

```python
type(serialized_pikachu)
```

```python
print(pikachu.model_dump_json(indent=3))
```

```python
with open("pikachu.json", "w") as file:
    file.write(pikachu.model_dump_json(indent=3))
```

```python
with open("pokemons.json", "w") as file:
    file.write(pokemons.model_dump_json(indent=3))
```
