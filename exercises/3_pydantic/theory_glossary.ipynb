## 4. Theory questions

- a) Explain the primary purpose of Pydantic. What core problems does it solve for developers?
    - It allows for easy validation with less steps, and clearer error messages, makes it easy to validate json-strings and save model into json object for use in another api, easy to filter out the fields you want to use from an api call, and exclude the rest. converts easily between python and json object, ensures consistency
    - Pydantic’s primary purpose is data validation and parsing using Python type hints.
    - It solves problems such as:
        - Making sure data matches the expected types and constraints.
        - Providing clear and structured error messages when validation fails.
        - Converting between JSON and Python objects with minimal boilerplate.
        - Enforcing consistency when dealing with external/untrusted data (e.g., from APIs).

- b) In Pydantic, what is the key difference between using model_validate() and model_validate_json()? When would you use each one?
    - model_validate() - validates a python objekt, (list, dict, list[dict])
    - model_validate_json() - validates a json-string

- c) How can you define a field in a Pydantic model that has a specific value constraint, such as a minimum or maximum number? Give an example.
    - you can use pydantics Field function where you can for example validate that the values are within an specific expected range
    - used for applying constraints on a models field
    - example: age: int = Field(gt=-1, lt=126) - age has to be of type(int) and the values has to be between 0-125

- d) Describe how Pydantic's data validation can improve the robustness and reliability of an application that processes data from an external API.
    - you can choose what to call the fields based on what you want to get from the API
    - Validates incoming API data to ensure it matches the expected schema.
    - Prevents crashes caused by missing or malformed fields.
    - Converts and standardizes data types (e.g., "42" → int).
    - Provides automatic error handling with clear messages.

- e) What is a computed field in Pydantic?
    - @computed_field is a decorator function that gives a property more functionality, like computing from another field to create a new field
    - the computed field is automatically computed from other fields in the model, and is not stored as its own field

- f) What are tokens in LLMs?
    - the ammount of "data"/information(textunits, ca 0,75 tokens per word) you can use within a timeframe, to not overload the LLM
    - Tokens are the basic units of text (pieces of words, subwords, or symbols) used by LLMs.
    - Example: "chatting" might be split into "chat" + "ting".
    - On average, 1 token ≈ ¾ of a word in English.
    - Token limits control how much input/output the model can handle at once.

- g) Large language models (LLMs) can struggle to generate a large amount of structured data, like 100 JSON objects in a single request, primarily due to token limitations and context window constraints.
    - context window = memory for an LLM within a conversation.
    - if you exceed the token limit, the llm can skip parts of the memory from the conversation(context window constraint)
    - Context window = maximum tokens the model can “remember” in one request.
    - If the request + response exceed this limit, the model loses earlier context.
    - Large structured data (like 100 JSON objects) can easily hit this limit, causing truncation or incomplete output.

- h) What are some strategies you can use to make an LLM's output more reliable when you need it to produce a specific format, such as JSON or Markdown?
    - to be specific and tell the LLM to give output in a specific format, and give an example for how the output should look
    - Be explicit in instructions: “Output only valid JSON”.
    - Provide an example of the desired format.
    - Use few-shot examples to guide formatting.
    - Ask the model to wrap output in code fences (```json).
    - Post-process with a JSON validator or parser to catch errors.

- i) What are few shot examples?
    - examples of how output should look like, and how it should be structured, with format, dict and list symbols, rules of how to behave and act.
    - Few-shot learning = giving the model examples of input → output pairs so it learns the pattern you want.
    - Example: If you want JSON responses, show it 2–3 small JSON examples before asking for more.

## Glossary

| terminology        | explanation |
| ------------------ | ----------- |
| few shot learning  | Training/instructing an LLM by giving a few examples of desired input → output pairs.            |
| token              | A chunk of text (word, subword, or symbol) that LLMs process; ~0.75 words per token.            |
| context window     | The maximum number of tokens an LLM can "remember" at once (input + output). memory for a conversation with and LLM            |
| pydantic           | A Python library for validating, parsing, and serializing data using type hints. validates, structures and dumps data in a easy way             |
| BaseModel          | The core Pydantic class that user models inherit from to gain validation & parsing features. a class that our class inherits from, so that our new class can use the BaseModels methods           |
| Field              | Function to add metadata and validation constraints (min/max, regex, etc.) to model fields. can set limitations and rules for our field            |
| model_fields       | Dictionary describing a model’s fields, their types, and constraints. dict for the model, showing the models fields and their type            |
| computed_field     | A property-like field that is dynamically computed from other fields. computes an existing field to create a new field            |
| model_validate     | Validates a Python object (dict/list) against the model schema. validates the input for the model            |
| model_dump         | Exports the model’s data as a Python object (e.g., dict). outputs a python object of the model            |
| type hinting       | used for showing what type each parameter should be in a function call. Specifying the expected data type of variables, function params, or returns.            |
| serialization      | converting python object into json-object            |
| deserialization    | converting json-object into python object            |
| prompt engineering | The practice of designing precise prompts to guide LLM behavior/output. writing a prompt that is clear and gives you minimal errors based on what you want.             |
| zero-shot learning | Getting an LLM to perform a task without providing any examples—only clear instructions.            |
