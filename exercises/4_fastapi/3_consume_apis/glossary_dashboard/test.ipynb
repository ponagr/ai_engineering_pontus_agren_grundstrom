```python
import streamlit as st
import json
import requests
import pandas as pd

# BASE_URL = "http://127.0.0.1:8000"

def read_glossary(endpoint):
    BASE_URL = "http://127.0.0.1:8000"
    response = requests.get(BASE_URL + endpoint).text
    data = json.loads(response)
    
    return pd.DataFrame(data)

# endpoint = "/glossary"
df = read_glossary("/glossary")
df
```
