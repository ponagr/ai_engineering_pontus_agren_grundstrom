## 4. Theory questions

- a) What is a REST API?
    - REST(Representational State Transfer)
    - an api layer used to structure communication between client and server over http
    - based on resources/endpoints(/users) and the CRUD operations(GET,POST,PUT,DELETE)

- b) What is CRUD?
    - stands for Create, Read, Update, Delete
    - the 4 basic operations used in APIs and databases to manage resources

- c) Why would you want to use an API, can't you just create everything in for example a streamlit app?
    - provides a standardized layer for communication between frontend and backend
    - can be reused by multiple frontends
    - improves security, since the backend is not directly exposed
    - easier to scale and maintain, since the different layers (frontend/backend) is not depending on eachother

- d) What does it mean with decoupling and tightly coupling?
    - Decoupling:
        - the different parts of the system(frontend, backend, database) are independent
        - can be changed or replaced without breaking the whole system
    - Tightly coupling:
        - the system parts are strongly dependent on eachother
        - makes it hard to change or reuse the system

- e) What are some other frameworks to create APIs in Python?
    - Flask, Django, Tornado, Falcon

- f) How would you mix Python together with a JavaScript frontend?
    - by setting up an API layer with different endpoints in python
    - where the frontend can send http requests to the API
    - this makes the backend and frontend decoupled


## Glossary

| terminology     | explanation |
| --------------- | ----------- |
| uvicorn         | a server to run fastapi async apps, making the API accessible over http           |
| endpoint        | a path after base url where a specific operation is performed. ex: '/name'          |
| path parameter  | a variable embedded in the url path, that can be accessed and used as a parameter in the API logic. ex: '/name/{id}'           |
| query parameter | a parameter that can be used for filtering/searching, in the url after an endpoint '?' followed by parameters: url/endpoint?limit=10          |
| routes          | the mapping between a url-path(endpoint) and the function handling the request            |
| request         | a http-request from the client (GET,POST,PUT,DELETE) sent to an API endpoint            |
| REST            | architectural style for APIs, based on resources and standard HTTP methods            |
| CRUD            | Create, Read, Update, Delete - the basic operations for resources           |
| put             | Update/replace existing data in the API (http-method)           |
| post            | Add/create new data in API (http-method)           |
| read            | operation to fetch data in the API using a 'get' request         |
| update          | Update/change existing data in the API with 'put' request               |
| field_validator | a pydantic-function to validate a single field in a model             |
| Field           | a pydantic function to add metadata and constraints to model fields            |
| response        | the servers response to a http request, containing information (data, status, headers) based on the request         |
| swagger ui      | a web-based ui '/docs' in FastAPI to explore and test API endpoints         |
| OpenAPI         | an open standard for describing REST APIs in a human readable format(json/yaml)            |
| curl            | a command-line tool to send http-requests to APIs from the terminal            |
