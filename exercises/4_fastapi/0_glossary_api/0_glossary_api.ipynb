## 0. FastAPI glossary API

Read the data from this repository called `fastapi_glossary.json`. Create Pydantic model(s) of the data in a separate script called `data_processing.py`.


```python
import json

with open("fastapi_glossary.json", "r") as f:
    data = f.read()
data = json.loads(data)

from pydantic import BaseModel, ValidationError

class Glossary(BaseModel):
    id: int
    word: str 
    meaning: str

glos_list = []

for d in data:
    try:
        glos_list.append(Glossary.model_validate(d))
    except ValidationError as err:
        print(err)
```


e) Test out your API using requests inside of a Jupyter notebook or a separate Python script. Try the different request types.

```python
import requests

BASE_URL = "http://127.0.0.1:8000"

response = requests.get(f"{BASE_URL}/glossary").json()
response
```

```python
response = requests.get(f"{BASE_URL}/glossary/", params={"word": "uvicorn"})
response.json()
```

```python
new_glossary = {"id": 20, "word": "model_dump", "meaning": "Exports the models data as a Python object (e.g., dict). outputs a python object of the model."}
response = requests.post(f"{BASE_URL}/glossary/add_glossary", json=new_glossary)
response.json()
```

```python
updated_glossary = {"id": 20, "word": "BaseModel", "meaning": "The core Pydantic class that user models inherit from to gain validation & parsing features. a class that our class inherits from, so that our new class can use the BaseModels methods"}
response = requests.put(f"{BASE_URL}/glossary/update_glossary", json=updated_glossary)
response.json()
```
