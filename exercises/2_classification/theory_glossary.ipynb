## Theory

- a) What is supervised learning and give 2-3 examples of supervised learning?    
    - when you train the model on known answers(target/label data), and can compare the correct answer with prediction to make the model better
    - e.g. house prices, miles per gallons and loan default prediction using regression or classification

- b) What are the differences between classification and regression?      
    - regression predicts continous numerical values, like house prices
    - classification predicts categorical labels or classes, a discrete value

- c) Why split into train/validation/test sets?       
    - train set: used to train the model
    - validation set: to be able to use val to choose best k(tune hyperparameters) for the model and data while preventing risk of overfitting and data leakage by not giving the model the real test data that will be used in the final model after k is chosen
    - test set: used to evaluate the model on unseen(untrained) data

- d) How does feature scaling impact KNN?     
    - it is very important, since knn uses distance, all data needs to be within the same values(range)
    - Scaling ensures all features contribute equally to distance calculations; without it, features with larger ranges dominate, which can bias predictions.

- e) How to find a k value for KNN?       
    - use val data and predict on multiple k-values for the model, evaluate with accuracy_score and choose the k with best accuracy  

- f) What are common metrics for evaluating classifcation?        
    - confusion_matrix, ConfusionMatrixDisplay, classification_report and accuracy_score(for deciding value of k with elbow plot)

- g) A fire alarm has 98% accuracy, but gets false positives, how could this impact if it is a good fire alarm or not?        
    - if a fire alarm goes of when there is no fire, it can make people not react when there actually is a fire, not good but better than a fire alarm that doesnt go off at all when there is a fire

- h) For a covid quick screening test, what metric out of accuracy, recall and precision would you choose and why?        
    - recall is important, since we dont want to risk sending someone with covid home to infect others, while precision only leads to extra caution, not danger

## Glossary

Fill in this table either by copying this into your own markdown file or copy it into a spreadsheet if you feel that is easier to work with.

| terminology    | explanation |
| -------------- | ----------- |
| target         | the answer and what we want the model to predict            |
| training set   | data for training the model to find and learn patterns and make predictions based of           |
| validation set | data for tuning hyperparameters and selecting the best model without risking data leakage and overfitting in the model            |
| test set       | unseen data used for evaluating the models performance           |
| recall         | used for evaluating proportions of FN(False Negatives) in predictions (focused on minimizing FN)      |
| f1 score       | overall model performance by balancing FP and FN           |
| regularization | discourages the model from relying too heavily on any single feature, reducing overfitting and improving generalization, e.g. by penalizing large weights - Adds a penalty during model training to limit the size of the model’s weights (e.g., L1/Lasso or L2/Ridge). The features stay in the model, but the model is discouraged from relying too heavily on any one feature.         |
| hyperparameter | a parameter set before training whose best value is determined using validation data             |
| overfitting    | when the model learns the training data too well, including noise, and performs poorly on new data          |
| underfitting   | when the model is too simple or trained too little, failing to capture patterns in the data          |
| feature        | data used to find patterns and predict the target            |
| precision      | used for evaluating proportions of FP(False Positives) in predictions (focused on minimizing FP)           |
